### Clustering Functions


```python
def run_kmeans_clustering(data, k_range=K_RANGE):
    """Run K-Means clustering for multiple k values.
    
    Args:
        data: Feature matrix [n_samples × n_features]
        k_range: List of k values to test
        
    Returns:
        dict: {k: {'labels': ..., 'silhouette': ..., 'kmeans': ...}}
    """
    results = {}
    
    for k in k_range:
        kmeans = KMeans(n_clusters=k, random_state=RANDOM_STATE)
        labels = kmeans.fit_predict(data)
        
        # Calculate silhouette score
        if k > 1:
            sil_score = silhouette_score(data, labels)
        else:
            sil_score = 0.0
        
        results[k] = {
            'labels': labels,
            'silhouette': sil_score,
            'kmeans': kmeans
        }
    
    return results

print("✓ run_kmeans_clustering() defined")
```

```python
def characterize_clusters(df, cluster_labels, stimuli=STIMULI):
    """Characterize each cluster with statistics.
    
    Args:
        df: DataFrame with neuron data (must have fish_id, neuron_id, area, valence)
        cluster_labels: Array of cluster assignments
        stimuli: List of stimulus columns
        
    Returns:
        DataFrame with cluster statistics
    """
    df = df.copy()
    df['cluster'] = cluster_labels
    
    cluster_stats = []
    
    for cluster_id in sorted(df['cluster'].unique()):
        cluster_neurons = df[df['cluster'] == cluster_id]
        
        # Basic stats
        n_neurons = len(cluster_neurons)
        
        # Mean response profile
        mean_responses = cluster_neurons[stimuli].mean()
        
        # Valence composition (if valence column exists)
        if 'valence' in df.columns:
            n_positive = (cluster_neurons['valence'] == 1).sum()
            n_negative = (cluster_neurons['valence'] == 0).sum()
            pct_positive = n_positive / n_neurons * 100 if n_neurons > 0 else 0
        else:
            n_positive = n_negative = pct_positive = np.nan
        
        # Brain region distribution
        top_areas = cluster_neurons['area'].value_counts().head(3).to_dict()
        
        # Fish distribution
        fish_counts = cluster_neurons['fish_id'].value_counts().to_dict()
        
        stats = {
            'cluster': cluster_id,
            'n_neurons': n_neurons,
            'pct_positive': pct_positive,
            'n_positive': n_positive,
            'n_negative': n_negative,
            'top_areas': top_areas,
            'fish_distribution': fish_counts
        }
        
        # Add mean responses
        for stim in stimuli:
            stats[f'mean_{stim}'] = mean_responses[stim]
        
        cluster_stats.append(stats)
    
    return pd.DataFrame(cluster_stats)

print("✓ characterize_clusters() defined")
```

### Visualization Functions


```python
def plot_pca_clusters(pca_result, labels, k, silhouette, explained_var, title=""):
    """Plot PCA scatter colored by cluster.
    
    Args:
        pca_result: PCA transformed data [n_samples × 2]
        labels: Cluster labels
        k: Number of clusters
        silhouette: Silhouette score
        explained_var: Explained variance ratio [PC1, PC2]
        title: Plot title
    """
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # Color by cluster
    colors = [CLUSTER_COLORS.get(l, 'gray') for l in labels]
    ax.scatter(pca_result[:, 0], pca_result[:, 1], c=colors, s=30, alpha=0.7)
    
    ax.set_xlabel(f"PC1 ({explained_var[0]:.1f}%)", fontsize=14)
    ax.set_ylabel(f"PC2 ({explained_var[1]:.1f}%)", fontsize=14)
    ax.set_title(f"{title}\nk = {k}, Silhouette = {silhouette:.3f}", fontsize=16)
    ax.tick_params(labelsize=12)
    
    plt.tight_layout()
    return fig

print("✓ plot_pca_clusters() defined")
```

```python
def plot_pca_by_valence(pca_result, df, explained_var):
    """Plot PCA colored by valence and correlation strength.
    
    Only works if 'valence' and 'correlation' columns exist.
    
    Args:
        pca_result: PCA transformed data
        df: DataFrame with valence and correlation columns
        explained_var: Explained variance ratio
    """
    if 'valence' not in df.columns or 'correlation' not in df.columns:
        print("Skipping valence plot - requires 'valence' and 'correlation' columns")
        return None
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # Color maps
    cmap_pos = LinearSegmentedColormap.from_list('pos', ['lightgreen', 'darkgreen'])
    cmap_neg = LinearSegmentedColormap.from_list('neg', ['pink', 'magenta'])
    norm = Normalize(vmin=0, vmax=0.8)
    
    # Separate by valence
    pos_mask = df['valence'] == 1
    neg_mask = df['valence'] == 0
    
    # Plot
    sc1 = ax.scatter(pca_result[pos_mask, 0], pca_result[pos_mask, 1],
                     c=df.loc[pos_mask, 'correlation'], cmap=cmap_pos, norm=norm,
                     s=40, alpha=0.9, label='Positive valence')
    
    sc2 = ax.scatter(pca_result[neg_mask, 0], pca_result[neg_mask, 1],
                     c=df.loc[neg_mask, 'correlation'], cmap=cmap_neg, norm=norm,
                     s=40, alpha=0.9, label='Negative valence')
    
    ax.set_xlabel(f"PC1 ({explained_var[0]:.1f}%)", fontsize=14)
    ax.set_ylabel(f"PC2 ({explained_var[1]:.1f}%)", fontsize=14)
    ax.set_title("PCA: Colored by Valence & Correlation", fontsize=16)
    ax.legend(fontsize=12)
    
    # Colorbars
    fig.colorbar(sc1, ax=ax, label='Correlation (positive)')
    fig.colorbar(sc2, ax=ax, label='Correlation (negative)')
    
    plt.tight_layout()
    return fig

print("✓ plot_pca_by_valence() defined")
```

```python
def plot_cluster_heatmap(cluster_summary, stimuli=STIMULI):
    """Plot heatmap of mean responses per cluster.
    
    Args:
        cluster_summary: DataFrame from characterize_clusters()
        stimuli: List of stimulus names
    """
    # Extract mean responses
    mean_cols = [f'mean_{s}' for s in stimuli]
    heatmap_data = cluster_summary[mean_cols].values
    
    fig, ax = plt.subplots(figsize=(8, len(cluster_summary) * 0.8))
    
    im = ax.imshow(heatmap_data, cmap='RdBu_r', aspect='auto', vmin=0, vmax=1)
    
    # Labels
    ax.set_xticks(range(len(stimuli)))
    ax.set_yticks(range(len(cluster_summary)))
    ax.set_xticklabels(stimuli, rotation=45, ha='right', fontsize=12)
    ax.set_yticklabels([f"Cluster {c}" for c in cluster_summary['cluster']], fontsize=12)
    
    # Colorbar
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label('Normalized Response', fontsize=12)
    
    ax.set_title("Mean Response Profile per Cluster", fontsize=14, fontweight='bold')
    
    plt.tight_layout()
    return fig

print("✓ plot_cluster_heatmap() defined")
```

### Dynamic Cluster Color Assignment

**Two color modes:**

1. **Automatic (default):** Colors assigned based on cluster response profiles
   - Attractive-preferring clusters → Green shades
   - Aversive-preferring clusters → Pink/Magenta shades
   - Balanced/Mixed clusters → Yellow/Orange shades

2. **Manual:** Override with your own colors
   - Set `USE_MANUAL_COLORS = True`
   - Define colors in `MANUAL_CLUSTER_COLORS`

**Colors are applied to:**
- PCA scatter plots
- Brain spatial plots
- Response line plots
- All visualizations


```python
def assign_cluster_colors_by_response(cluster_summary, stimuli=STIMULI):
    """Automatically assign colors based on cluster response profiles.
    
    Logic:
    - Analyzes mean response to attractive vs aversive stimuli
    - Attractive-preferring → green shades
    - Aversive-preferring → pink/magenta shades
    - Balanced/Mixed → orange/yellow shades
    
    Args:
        cluster_summary: DataFrame with mean_{stimulus} columns
        stimuli: List of stimulus names
        
    Returns:
        dict: {cluster_id: color_string}
    """
    color_map = {}
    
    # Define color palettes
    attractive_colors = ['limegreen', 'forestgreen', 'green', 'darkgreen', 'seagreen']
    aversive_colors = ['magenta', 'deeppink', 'hotpink', 'orchid', 'mediumvioletred']
    mixed_colors = ['orange', 'gold', 'darkorange', 'coral', 'goldenrod']
    
    for idx, row in cluster_summary.iterrows():
        cluster_id = int(row['cluster'])
        
        # Calculate attractive vs aversive preference
        attr_responses = [row[f'mean_{s}'] for s in ATTRACTIVE if f'mean_{s}' in row]
        aver_responses = [row[f'mean_{s}'] for s in AVERSIVE if f'mean_{s}' in row]
        
        attr_mean = np.mean(attr_responses) if attr_responses else 0
        aver_mean = np.mean(aver_responses) if aver_responses else 0
        
        # Get overall response strength
        all_responses = [row[f'mean_{s}'] for s in stimuli if f'mean_{s}' in row]
        max_response = max(all_responses) if all_responses else 0
        
        # Assign color based on preference
        if max_response < 0.3:
            # Low overall response - mixed
            color = mixed_colors[idx % len(mixed_colors)]
        elif abs(attr_mean - aver_mean) < 0.15:
            # Balanced between attractive and aversive
            color = mixed_colors[idx % len(mixed_colors)]
        elif attr_mean > aver_mean:
            # Prefers attractive stimuli → green
            color = attractive_colors[idx % len(attractive_colors)]
        else:
            # Prefers aversive stimuli → pink/magenta
            color = aversive_colors[idx % len(aversive_colors)]
        
        color_map[cluster_id] = color
    
    return color_map

print("✓ assign_cluster_colors_by_response() defined")
```

```python
def describe_cluster_colors(cluster_summary, color_map, stimuli=STIMULI):
    """Print explanation of color assignments.
    
    Args:
        cluster_summary: DataFrame with cluster stats
        color_map: Dict {cluster_id: color}
        stimuli: List of stimulus names
    """
    print("\n" + "="*80)
    print("CLUSTER COLOR ASSIGNMENTS & RATIONALE")
    print("="*80)
    
    for _, row in cluster_summary.iterrows():
        cluster_id = int(row['cluster'])
        color = color_map.get(cluster_id, 'gray')
        
        # Calculate preferences
        attr_responses = [row[f'mean_{s}'] for s in ATTRACTIVE if f'mean_{s}' in row]
        aver_responses = [row[f'mean_{s}'] for s in AVERSIVE if f'mean_{s}' in row]
        
        attr_mean = np.mean(attr_responses) if attr_responses else 0
        aver_mean = np.mean(aver_responses) if aver_responses else 0
        
        # Find dominant stimulus
        mean_responses = {s: row[f'mean_{s}'] for s in stimuli if f'mean_{s}' in row}
        if mean_responses:
            max_stim = max(mean_responses, key=mean_responses.get)
            max_val = mean_responses[max_stim]
        else:
            max_stim = 'N/A'
            max_val = 0
        
        print(f"\nCluster {cluster_id}: {color.upper()}")
        print(f"  Dominant stimulus: {max_stim} (response: {max_val:.3f})")
        print(f"  Attractive mean: {attr_mean:.3f}")
        print(f"  Aversive mean: {aver_mean:.3f}")
        print(f"  Preference: ", end="")
        
        if abs(attr_mean - aver_mean) < 0.15:
            print("Balanced/Mixed")
        elif attr_mean > aver_mean:
            print(f"Attractive ({attr_mean - aver_mean:.3f} higher)")
        else:
            print(f"Aversive ({aver_mean - attr_mean:.3f} higher)")
    
    print("\n" + "="*80)

print("✓ describe_cluster_colors() defined")
```

```python
def plot_cluster_response_profiles(cluster_summary, color_map, stimuli=STIMULI):
    """Plot response profiles as line plots with cluster-specific colors.
    
    Args:
        cluster_summary: DataFrame with mean responses
        color_map: Dict {cluster_id: color}
        stimuli: List of stimulus names
    """
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # Plot each cluster
    for _, row in cluster_summary.iterrows():
        cluster_id = int(row['cluster'])
        color = color_map.get(cluster_id, 'gray')
        
        # Get mean responses
        responses = [row[f'mean_{s}'] for s in stimuli]
        
        # Plot line
        ax.plot(stimuli, responses, 
                color=color, 
                marker='o', 
                linewidth=2.5,
                markersize=8,
                label=f'Cluster {cluster_id} (n={int(row["n_neurons"])})',
                alpha=0.8)
    
    # Formatting
    ax.set_xlabel('Stimulus', fontsize=14, fontweight='bold')
    ax.set_ylabel('Normalized Response', fontsize=14, fontweight='bold')
    ax.set_title('Cluster Response Profiles (Colored by Preference)', 
                fontsize=16, fontweight='bold')
    ax.set_ylim(-0.05, 1.05)
    ax.grid(alpha=0.3, linestyle='--')
    ax.legend(fontsize=11, loc='best')
    
    # Rotate x labels
    plt.xticks(rotation=45, ha='right', fontsize=12)
    plt.yticks(fontsize=12)
    
    # Add vertical line separating attractive/aversive
    if len(ATTRACTIVE) > 0:
        ax.axvline(x=len(ATTRACTIVE)-0.5, color='gray', linestyle='--', 
                  alpha=0.5, linewidth=1.5)
        ax.text(len(ATTRACTIVE)/2, 1.02, 'Attractive', 
               ha='center', fontsize=11, style='italic')
        ax.text(len(ATTRACTIVE) + (len(AVERSIVE))/2, 1.02, 'Aversive',
               ha='center', fontsize=11, style='italic')
    
    plt.tight_layout()
    return fig

print("✓ plot_cluster_response_profiles() defined")
```

```python
def plot_brain_spatial(ref_brain, coords, labels, title=""):
    """Plot neurons on brain atlas colored by cluster.
    
    Args:
        ref_brain: 3D reference brain image
        coords: Neuron coordinates [n_neurons × 3]
        labels: Cluster labels
        title: Plot title
    """
    fig = plt.figure(figsize=(10, 12))
    gs = gridspec.GridSpec(2, 1, height_ratios=[1, 1])
    gs.update(hspace=-0.6)
    
    ax_horizontal = fig.add_subplot(gs[0])
    ax_sagittal = fig.add_subplot(gs[1])
    
    unique_labels = np.unique(labels)
    
    # Horizontal projection (top view)
    proj_horizontal = np.nanmean(ref_brain, axis=0)
    proj_horizontal = np.rot90(proj_horizontal, k=1)
    ax_horizontal.imshow(proj_horizontal, cmap='gray_r', alpha=0.5, origin='lower')
    
    for label in unique_labels:
        color = CLUSTER_COLORS.get(label, 'gray')
        label_coords = coords[labels == label]
        ax_horizontal.scatter(
            label_coords[:, 1],
            proj_horizontal.shape[0] - label_coords[:, 2],
            c=color, s=20, alpha=0.9, label=f'Cluster {label}'
        )
    
    ax_horizontal.axis('off')
    ax_horizontal.set_aspect('equal')
    ax_horizontal.legend(loc='upper right', fontsize=10)
    
    # Sagittal projection (side view)
    proj_sagittal = np.nanmean(ref_brain, axis=2)
    ax_sagittal.imshow(proj_sagittal, cmap='gray_r', alpha=0.5, origin='lower')
    
    for label in unique_labels:
        color = CLUSTER_COLORS.get(label, 'gray')
        label_coords = coords[labels == label]
        ax_sagittal.scatter(
            label_coords[:, 0],
            label_coords[:, 1],
            c=color, s=20, alpha=0.9
        )
    
    ax_sagittal.axis('off')
    ax_sagittal.set_aspect('equal')
    
    if title:
        fig.suptitle(title, fontsize=16, fontweight='bold', y=0.98)
    
    plt.tight_layout()
    return fig

print("✓ plot_brain_spatial() defined")
print("\n" + "="*80)
print("All helper functions loaded!")
print("="*80)
```

## 3. Data Loading

**Mode-dependent neuron selection:**
- `'combined'`: Load both positive and negative valence neurons
- `'positive_only'`: Load only positive valence neurons  
- `'negative_only'`: Load only negative valence neurons

All neurons filtered by p_value < 0.05


### Load Valence Correlation Data


```python
# Load positive and negative valence neurons
pos_regress = pd.read_csv(POS_VALENCE_PATH)
neg_regress = pd.read_csv(NEG_VALENCE_PATH)

print(f"Loaded valence data:")
print(f"  Positive neurons: {len(pos_regress)}")
print(f"  Negative neurons: {len(neg_regress)}")

# Add valence labels
pos_regress['valence'] = 1  # Positive valence
neg_regress['valence'] = 0  # Negative valence
```

### Select Neurons Based on MODE


```python
# Select neurons based on MODE
if MODE == 'positive_only':
    valence = pos_regress[pos_regress['p_value'] < P_VALUE_THRESHOLD].copy()
    mode_description = "Positive valence neurons only"
    
elif MODE == 'negative_only':
    valence = neg_regress[neg_regress['p_value'] < P_VALUE_THRESHOLD].copy()
    mode_description = "Negative valence neurons only"
    
else:  # combined
    valence = pd.concat([
        pos_regress[pos_regress['p_value'] < P_VALUE_THRESHOLD],
        neg_regress[neg_regress['p_value'] < P_VALUE_THRESHOLD]
    ], ignore_index=True)
    mode_description = "Combined positive + negative valence neurons"

print(f"\n{'='*80}")
print(f"MODE: {MODE}")
print(f"{'='*80}")
print(f"Description: {mode_description}")
print(f"Selected neurons: {len(valence)}")
if MODE == 'combined':
    n_pos = (valence['valence'] == 1).sum()
    n_neg = (valence['valence'] == 0).sum()
    print(f"  Positive: {n_pos} ({n_pos/len(valence)*100:.1f}%)")
    print(f"  Negative: {n_neg} ({n_neg/len(valence)*100:.1f}%)")
print(f"{'='*80}")
```

### Load Neural Response Data


```python
# Load full neural response dataset
data_resp = pd.read_hdf(NEURAL_DATA_PATH)
data_resp.rename(columns={"coords_seralised": "coords"})

print(f"\nLoaded neural responses: {len(data_resp)} records")

# Merge valence neurons with their response data
stim_columns = valence[['correlation', 'coords', 'p_value', 'valence']]
merged_df = pd.merge(stim_columns, data_resp, on=['coords'], how='inner')

print(f"Merged data: {len(merged_df)} records")

# Filter for our 6 stimuli
conc_df = merged_df[merged_df['stimulus'].isin(STIMULI)]
print(f"Filtered for {len(STIMULI)} stimuli: {len(conc_df)} records")
```

### Average and Pivot Responses


```python
# Average responses across trials
averaged_responses = merged_df.groupby(
    ['fish_id', 'neuron_id', 'stimulus', 'coords', 'area', 'valence']
)['resp'].mean().reset_index()

print(f"Averaged responses: {len(averaged_responses)} records")

# Sort by stimulus for consistency
averaged_responses['stimulus'] = pd.Categorical(
    averaged_responses['stimulus'], 
    categories=STIMULI, 
    ordered=True
)
averaged_responses = averaged_responses.sort_values(['fish_id', 'neuron_id', 'stimulus'])

# Pivot: neurons × stimuli
pivot_df = averaged_responses.pivot_table(
    index=['fish_id', 'neuron_id', 'coords', 'area', 'valence'],
    columns='stimulus',
    values='resp'
).reset_index()

# Drop neurons with any missing responses
pivot_df = pivot_df.dropna(subset=STIMULI)

print(f"\nFinal neuron matrix: {len(pivot_df)} neurons × {len(STIMULI)} stimuli")
print(f"\n✓ Data loading complete")
```

### Normalize Responses


```python
# Extract response matrix
responses = pivot_df[STIMULI].values

# MinMax scale per neuron (row-wise)
# Each neuron's responses scaled to 0-1 range
scaled_responses = np.array([
    MinMaxScaler().fit_transform(r.reshape(-1, 1)).flatten() 
    for r in responses
])

# Create DataFrame
scaled_df = pd.DataFrame(scaled_responses, columns=STIMULI, index=pivot_df.index)

# Merge back with metadata
scaled_df = scaled_df.reset_index()
metadata_df = pivot_df[['fish_id', 'neuron_id', 'coords', 'area', 'valence']].reset_index(drop=True)
scaled_df = pd.concat([metadata_df, scaled_df[STIMULI]], axis=1)

print(f"✓ Responses normalized (MinMax per neuron)")
print(f"  Shape: {scaled_responses.shape}")
print(f"  Range: [{scaled_responses.min():.2f}, {scaled_responses.max():.2f}]")
```

## 4. Clustering Analysis

Run K-Means for multiple k values and evaluate with silhouette scores


### Run K-Means Clustering


```python
# Run clustering for all k values
clustering_results = run_kmeans_clustering(scaled_responses, k_range=K_RANGE)

print(f"\n{'='*80}")
print("CLUSTERING RESULTS")
print(f"{'='*80}")
for k, result in clustering_results.items():
    print(f"k={k}: Silhouette = {result['silhouette']:.3f}")
print(f"{'='*80}")

# Find optimal k (highest silhouette)
optimal_k = max(clustering_results.keys(), 
                key=lambda k: clustering_results[k]['silhouette'])
print(f"\n✓ Optimal k (highest silhouette): {optimal_k}")
```

### PCA for Visualization


```python
# PCA to 2D for visualization
pca = PCA(n_components=2, random_state=RANDOM_STATE)
pca_result = pca.fit_transform(scaled_responses)
explained_var = pca.explained_variance_ratio_ * 100

print(f"\nPCA completed:")
print(f"  PC1 explains: {explained_var[0]:.1f}%")
print(f"  PC2 explains: {explained_var[1]:.1f}%")
print(f"  Total: {explained_var.sum():.1f}%")

# Add PCA to dataframe
scaled_df['PC1'] = pca_result[:, 0]
scaled_df['PC2'] = pca_result[:, 1]

# Add correlation info for visualization
if 'correlation' in valence.columns:
    corr_map = valence.set_index('coords')['correlation'].to_dict()
    scaled_df['correlation'] = scaled_df['coords'].apply(
        lambda c: corr_map.get(c, np.nan)
    )
```

### Visualize Clustering Results


```python
# Plot PCA for all k values
fig, axes = plt.subplots(1, len(K_RANGE), figsize=(6*len(K_RANGE), 5))
if len(K_RANGE) == 1:
    axes = [axes]

for i, k in enumerate(K_RANGE):
    ax = axes[i]
    result = clustering_results[k]
    
    # Color by cluster
    colors = [CLUSTER_COLORS.get(l, 'gray') for l in result['labels']]
    ax.scatter(pca_result[:, 0], pca_result[:, 1], c=colors, s=30, alpha=0.7)
    
    ax.set_xlabel(f"PC1 ({explained_var[0]:.1f}%)", fontsize=14)
    ax.set_ylabel(f"PC2 ({explained_var[1]:.1f}%)", fontsize=14)
    ax.set_title(f"k = {k}\nSilhouette = {result['silhouette']:.3f}", fontsize=16)
    ax.tick_params(labelsize=12)

plt.suptitle(f"Clustering Results - MODE: {MODE}", fontsize=18, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n✓ PCA clustering plots displayed")
```

### PCA Colored by Valence (Combined Mode Only)


```python
# Only plot if in combined mode
if MODE == 'combined' and 'correlation' in scaled_df.columns:
    fig = plot_pca_by_valence(pca_result, scaled_df, explained_var)
    if fig:
        plt.show()
        print("\n✓ Valence-colored PCA plot displayed")
else:
    print(f"\nSkipping valence plot (MODE = {MODE})")
```

### Cluster Composition (Forebrain vs Hindbrain)


```python
# Extract fish type (FB vs HB)
scaled_df['fish_type'] = scaled_df['fish_id'].apply(
    lambda x: 'Forebrain' if '_fb' in x else 'Hindbrain'
)

# Count neurons per cluster per fish type
comp_counts = scaled_df.groupby(['fish_type', 'cluster']).size().reset_index(name='count')

# Normalize to fractions
comp_totals = comp_counts.groupby('fish_type')['count'].transform('sum')
comp_counts['fraction'] = comp_counts['count'] / comp_totals

# Pivot for plotting
comp_pivot = comp_counts.pivot(index='fish_type', columns='cluster', values='fraction').fillna(0)

# Plot stacked bar chart
fig, ax = plt.subplots(figsize=(8, 6))

# Get colors in order
cluster_order = sorted(comp_pivot.columns)
colors = [CLUSTER_COLORS.get(c, 'gray') for c in cluster_order]

comp_pivot[cluster_order].plot(kind='bar', stacked=True, color=colors, ax=ax)

ax.set_ylabel('Fraction of Neurons', fontsize=14)
ax.set_xlabel('')
ax.set_xticklabels(['Forebrain', 'Hindbrain'], rotation=0, fontsize=14)
ax.set_title(f'Cluster Composition by Brain Region - MODE: {MODE}', 
             fontsize=16, fontweight='bold')
ax.legend(title='Cluster', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

print("\n✓ Cluster composition plot displayed")
```
